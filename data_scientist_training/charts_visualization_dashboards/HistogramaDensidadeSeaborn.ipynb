Importação das bibliotecas

```python
import pandas as pd
import seaborn as srn
import matplotlib.pyplot as plt
```

Carregamento da base da dados

```python
base = pd.read_csv('trees.csv')
base.head()
```

Histograma com 10 divisões(bins) e com gráfico de densidade

```python
# stat: count, frequency, probability, percent, density
srn.histplot(base.Volume, color='blue', kde=True, stat='density', bins=10).set(title='Volume')
```
