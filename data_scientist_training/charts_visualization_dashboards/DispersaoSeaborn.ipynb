Importação das bibliotecas

```python
import pandas as pd
import seaborn as srn
import matplotlib.pyplot as plt
```

Caregamento da base de dados

```python
base = pd.read_csv('co2.csv')
base.head()
```

Gráfico de dispersão utlizando os atributos conc e uptake, agrupamento pelo type

```python
srn.scatterplot(data=base, x='conc', y='uptake', hue='Type')
plt.show()
```

Seleção de registros especificos da base de dados (Quebec e Mississipi)

```python
q = base.loc[base['Type'] == 'Quebec']
m = base.loc[base['Type'] == 'Mississippi']
```

Subgrafico (1 linha e duas colunas) mostrando gráficos sobre cada região

```python
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.title('Quebec')
srn.scatterplot(x=q.conc, y=q.uptake)
plt.subplot(1, 2, 2)
srn.scatterplot(x=m.conc, y=m.uptake)
plt.title('Mississippi')
plt.tight_layout()
plt.show()
```

Refrigerado e não refrigerado

```python
ch = base.loc[base['Treatment'] == 'chilled']
nc = base.loc[base['Treatment'] == 'nonchilled']
```

Gráfico somente com 'chilled' e 'nonchilled'

```python
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
srn.scatterplot(x=ch.conc, y=ch.uptake)
plt.title('Chilled')
plt.subplot(1, 2, 2)
srn.scatterplot(x=nc.conc, y=nc.uptake)
plt.title('Non chilled')
plt.tight_layout()
plt.show()
```

Carregamento de outro arquivo, cancer de esofago

```python
base2 = pd.read_csv('esoph.csv')
base2
```

Gráfico entre os atributos 'alcgp' e 'ncontrols'

```python
srn.catplot(x = 'alcgp', y = 'ncontrols', data = base2, jitter = False)
plt.show()
```

Gráfico entre os atributos 'alcgp' e 'ncontrols', com agrupamento

```python
srn.catplot(x = 'alcgp', y = 'ncontrols', data = base2, col = 'tobgp')
plt.show()
```
