Importação da função norm

```python
from scipy.stats import norm
```

Conjunto de objetos em uma cesta, a média é 8 e o desvio padrão é 2

Qual a probabilidade de tirar um obejato que o peso é menor que 6 quilos?

```python
norm.cdf(6, 8, 2)
```


Qual a probabilidade de tirar um obejato que o peso é maior que 6 quilos?

```python
norm.sf(6, 8, 2)
# 1 - norm.cdf(6, 8, 2)
```

Qual a probabilidade de tirar um obejato que o peso é menor que 6 ou maior que 10 quilos?

```python
norm.cdf(6, 8, 2) + norm.sf(10, 8, 2)
```

Qual a probabilidade de tirar um obejato que o peso é menor que 10 ou maior que 8 quilos?

```python
norm.cdf(10, 8, 2) - norm.cdf(8, 8, 2)
```
