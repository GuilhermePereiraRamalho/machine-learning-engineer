Importação das bibliotecas

```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

Carregamento da base de dados

```python
iris = pd.read_csv('iris.csv')
iris['class'].value_counts()
```

```python
x, _, y, _ = train_test_split(
  iris.iloc[:,:4],
  iris.iloc[:, 4],
  test_size=0.5,
  stratify=iris.iloc[:, 4]
)
y.value_counts()
```
