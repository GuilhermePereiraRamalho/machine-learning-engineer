```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import confusion_matrix, accuracy_score
from sklearn.ensemble import RandomForestClassifier
```

```python
credit = pd.read_csv('Credit.csv')
credit.head()
```

```python
labelencoder = LabelEncoder()
for i in credit.select_dtypes(include='object'):
  if i != 'class':
    credit[i] = labelencoder.fit_transform(credit[i])
```

```python
credit.head()
```

```python
previsores = credit.iloc[:,0:20].values
classe = credit.iloc[:,20].values
```

```python
previsores
```

```python
print(type(previsores))
```

```python
print(type(classe))
```

```python
x_train, x_test, y_train, y_test = train_test_split(previsores, classe, test_size=0.3, random_state=0)
```

```python
print(x_train.shape)
print(y_train.shape)
print(x_test.shape)
print(y_test.shape)
```

```python
forest = RandomForestClassifier(n_estimators=500, random_state=0)
forest.fit(x_train, y_train)
```

```python
previsoes = forest.predict(x_test)
print(previsoes)
```

```python
confusion_matrix = confusion_matrix(y_test, previsoes)
print(confusion_matrix)
```

```python
accuracy_score = accuracy_score(y_test, previsoes)
print(accuracy_score)
```
