# Basic Examples

```python
import re

text = '00123451'

info = re.search('1', text)

if info != None:
  print(f'Occurrence found in: {info.span()}')
  print(f'What was found: {info.group()}')
else:
  print("Nothing was found!")
```

```python
import re

text = '00123451'

info = re.findall('1', text)
print(info)
```

```python
import re

text = '001234510'

info = re.split("1", text)
print(info)
```

```python
import re

text = '001234510'

info = re.sub("1", "#", text)
print(info)
```

```python
import re

text = 'There are 64 buildings that are 700 meters tall'

info = re.search('buildings', text)

if info != None:
  print(f'Occurrence found in: {info.span()}')
  print(f'What was found: {info.group()}')
else:
  print("Nothing was found!")
```

```python
import re

text = 'ABCDefgHI123'

info1 = re.findall('[Ae3]', text)
info2 = re.findall('[A-Z]', text)
info3 = re.findall('[a-z]', text)
info4 = re.findall('[0-9]', text)
info5 = re.findall('[A-Za-z]', text)

print(info1)
print(info2)
print(info3)
print(info4)
print(info5)
```

4 - Phone numbers ending with 95 must be blocked. A phone number must be valid to be validated, in the format XXXXXXXX, where X is a digit. First, determine whether it is a valid number. If it is, determine whether it is blocked or not.

```python
import re

text = '95234567'

info = re.search("^([0-9]{8})$", text)

if info != None:
  print('valid number')
  info2 = re.search("^95([0-9]{6})$", text)
  if info2 != None:
    print("blocked phone number")
  else:
    print("unblocked phone number")
else:
  print("invalid number")
```

5 - Write a regular expression to recognize words ending in the gerund form. Normally, these words can be detected if they end with ando, endo, or indo. Examples: sorrindo, andando. Use the findall function to return the occurrences.

```python
import re

text = 'Olá, eu estou dormindo, e não sorrindo'

info = re.findall(r"([\w]+indo|ando|end)+", text)

print(info)
```

6 - Write a regular expression to determine whether a time is valid. The format uses a 24-hour clock and is specified as follows: HH:MM

Examples:

19:30
09:30
23:45
23:70 (invalid)

```python
import re

text = '19:30'

info = re.search("^([0-1][0-9]|[2][0-4]):([0-5][0-9])$", text)

if info != None:
  print("Valid hour")
else:
  print("Invalid hour")
```

7 (CHALLENGE) - Write a regular expression to validate whether an expression is a valid mathematical operation. The expression may contain only two integers, which must be added or subtracted. Validate whether the expression is a valid mathematical expression or not.

```python
import re

text = '310 + 4453'

info = re.search(r"^ *(\d+ *(-|\+) *\d+) *$",text)

if info != None:
  print("valid mathematical operation")
else:
  print("invalid mathematical operation")
```
