# Creating Our Types

```python
import numpy as np

class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age


array = np.array(
    [
        Person("Fernando", 45),
        Person("Rodrigo", 23)
    ]
)

print(array)
print(type(array))
print(array.dtype, end="\n\n")
print(array[0].name, array[1].age)
```

```python
import numpy as np

person_type = np.dtype([('name', 'S10'), ('age', 'i4')])

array = np.array(
    [
        ("Fernando", 45),
        ("Rodrigo", 23)
    ],
    dtype=person_type
)

print(array)
print(type(array))
print(array.dtype, end="\n\n")
```

Create a function that splits an array into N pieces, but keeps only the absolute values of the elements in the array.

```python
import numpy as np

def split_absolute(arr, n):
  arr = np.array_split(arr, n)
  arr = np.abs(arr)
  return arr

array1 = np.array([1,2,3,-4,5,-7])
print(split_absolute(array1, 2))
```

Create a function that removes negative numbers from an array.

```python
import numpy as np

def remove_negatives(arr):
  filter = arr >= 0
  return arr[filter]

array = np.array([1,-5,5,4,3,-2,-1])
print(remove_negatives(array))
```

Create a function that receives a starting number, an ending number, and an array. The function should return a new array containing only the numbers within this range.

```python
import numpy as np

def filter_values(arr, start, end):
  filter = (arr >= start) & (arr <= end)
  return arr[filter]

array = np.array([i for i in range(-10,11)])
print(array)

print(filter_values(array, 0, 7))
```

Create a function that sorts a matrix and removes all odd numbers.

```python
import numpy as np

def sort_evens(arr):
  arr = np.sort(arr)
  filter = arr % 2 ==1
  return arr[filter]

array = np.random.randint(0,10,(100))
print(sort_evens(array))
```

Do the same as the previous exercise, but also remove duplicate values.

```python
import numpy as np

def sort_unique_evens(arr):
  arr = np.sort(arr)
  filter = arr % 2 ==1
  return np.unique(arr[filter])

array = np.random.randint(0,10,(100))
print(sort_unique_evens(array))
```
