# Stacked Charts

```python
xs = np.array(['January', 'February', 'March', 'April'])
team1 = np.array([5,10,15,20])
team2 = np.array([1,2,4,9])
team3 = np.array([5,8,2,2])
team4 = np.array([6,7,2,1])

plt.bar(xs, team1, color='red', width=0.5)
plt.bar(xs, team2, bottom=team1, color='yellow', width=0.5)
plt.bar(xs, team3, bottom=team2+team1, color='orange', width=0.5)
plt.bar(xs, team4, bottom=team3+team2+team1, color='green', width=0.5)

plt.title("Sports Performance")
plt.xlabel("Months")
plt.ylabel("Goals")
plt.legend(['Team1', 'Team2', 'Team3', 'Team4'])

plt.show()
```
