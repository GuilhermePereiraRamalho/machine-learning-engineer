# Grouping Values

```python
import pandas as pd

data = pd.read_csv('pessoas.csv', index_col='Nome')
data
```

```python
data.loc["Willian"] = (25,"Programador", 1.74)
data.loc["Lando"] = (22,"Programador", 1.65)
data.loc["Max"] = (45, "Médico", 1.68)
data.loc["Sebastian"] = (35, "Médico", 1.74)
data.loc["Ana"] = (23, "Médico", 1.70)

data
```

```python
groups = data.groupby('Profissão')
```

```python
for i, group in groups:
  print(i)
  print(group)
```

```python
groups.get_group('Programador')
```

```python
groups = data.groupby(['Profissão', 'Idade'])
groups.describe()
```

```python
import pandas as pd

data = {
    'Sede': ['Nova Iorque', 'São Paulo', 'Nova Iorque', 'Londres', 'Londres'],
    'Piloto': ['Mike Ross', 'Sebastian Thomas', 'Glen Are', 'Michael Schum', 'Luiz da Silva'],
    'Mundiais': [10, 11, 0, 3, 44],
    'Vitórias': [321, 229, 12, 44, 1023]
}

data_frame = pd.DataFrame(data, index=["X Racing", "Equatorial", "Type", "Blue Race", "Capa Racing"])
data_frame
```

Display all information for the Blue Race team.

```python
data_frame.loc['Blue Race']
```

Display the number of championships won by the Capa Racing team.

```python
data_frame.loc['Capa Racing']['Mundiais']
```

Display only the teams that have 10 or more championships.

```python
mask = data_frame['Mundiais'] >= 10
data_frame[mask]
```

Display only the teams that have 10 or more championships
and more than 300 victories.

```python
mask = (data_frame['Mundiais'] >= 10) & (data_frame['Vitórias'] > 300)
data_frame[mask]
```
