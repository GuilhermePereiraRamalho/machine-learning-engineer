# Creating a Person Class with CSV Persistence

````python
import csv

class Person:
  def __init__(self, id, name, occupation):
    self.id = id
    self.name = name
    self.occupation = occupation

  @staticmethod
  def read_people():
    people = list()
    with open('people.csv', 'r') as file:
      reader = csv.reader(file, delimiter=',')
      for line in reader:
        person = Person(line[0], line[1], line[2])
        people.append(person)
    return people

  @staticmethod
  def save_people(*people):
    with open('people.csv', 'w') as file:
      csv_writer = csv.writer(file, delimiter=',')
      for person in people:
        csv_writer.writerow([person.id, person.name, person.occupation])

person1 = Person(23, 'Joseph', 'Engineer')
person2 = Person(12, 'Mary', 'Architect')
person3 = Person(44, 'Anna', 'Data Scientist')

Person.save_people(
    person1,
    person2,
    person3
)

people = Person.read_people()

for person in people:
  print(person.id, person.name, person.occupation)
````

````python
import xml.etree.ElementTree as xml
import os

def create_person_tag(name, cpf, sex, address):
  person_node = xml.Element("Person", attrib={'Name': name})

  cpf_node = xml.SubElement(person_node, 'CPF')
  cpf_node.text = cpf

  sex_node = xml.SubElement(person_node, 'Sex')
  sex_node.text = sex

  address_node = xml.SubElement(person_node, 'Address')
  address_node.text = address

  return person_node


data = {
    'Rodrigo': {
        'CPF': '123456788',
        'Sex': 'Male',
        'Address': 'Example 1',
        'Age': 23
    },
    'Fernando': {
        'CPF': '454545465',
        'Sex': 'Male',
        'Address': 'Example 2',
        'Age': 60,
        'Children': [
            'Rodrigo',
            'Lucas'
        ]
    },
    'Ana': {
        'CPF': '321321321',
        'Sex': 'Female',
        'Address': 'Example 3',
        'Age': 28
    }
}

root_node = xml.Element("PersonalData")
for key in data:
  name = key
  person_data = data[name]
  cpf = person_data['CPF']
  sex = person_data['Sex']
  address = person_data['Address']
  person = create_person_tag(name, cpf, sex, address)
  if 'Children' in person_data:
    children = xml.SubElement(person, 'Children')
    for child in person_data['Children']:
      child_node = xml.SubElement(children, 'Person', attrib={'name': child})
  root_node.append(person)

tree = xml.ElementTree(root_node)
with open('data_example3.xml', 'wb') as files:
  tree.write(files)
````

````python
import xml.etree.ElementTree as xml
import os

class Car:
  def __init__(self, name, power):
    self.name = name
    self.power = power

  @staticmethod
  def save_cars(*cars):
    root = xml.Element("Root")

    for car in cars:
      car_node = xml.Element('Car')

      name_node = xml.SubElement(car_node, 'Name')
      name_node.text = car.name

      power_node = xml.SubElement(car_node, 'Power')
      power_node.text = str(car.power)

      root.append(car_node)

    tree = xml.ElementTree(root)
    with open('cars.xml', 'wb') as files:
      tree.write(files)

  @staticmethod
  def read_cars():
    cars = list()
    tree = xml.parse('cars.xml')
    root = tree.getroot()

    for car in root:
      name = None
      power = None

      for data in car:
        if (data.tag == 'Name'):
          name = data.text
        if (data.tag == 'Power'):
          power = data.text
      car = Car(name, power)
      cars.append(car)
    return cars

car1 = Car('Maverik', 330)
car2 = Car('Nivus', 200)
car3 = Car('Monza', 180)

Car.save_cars(car1, car2, car3)

cars = Car.read_cars()

for car in cars:
  print(car.name, car.power)
````

````python
class Product:
  def __init__(self, name, price):
    self.name = name
    self.price = price

products = list()

with open('exercicio2.txt', 'rt') as file:
  for line in file:
    split_index = line.index('R$')

    name = line[:split_index -1]
    price = line[split_index:len(line)-1]
    product = Product(name, price)
    products.append(product)

for product in products:
  print(product.name, product.price)
````

````python
import csv

class Company:
  def __init__(self, id, name, number_of_employees, profit):
      self.id = id
      self.name = name
      self.number_of_employees = number_of_employees
      self.profit = profit

companies = list()

with open('exercicio6.csv', 'r') as file:
  reader = csv.reader(file, delimiter=',')
  for line in reader:
    comp = Company(line[0], line[1], line[2], line[3])
    companies.append(comp)

for comp in companies:
  print(comp.id, comp.name, comp.number_of_employees, comp.profit)
````

7 - Create a class that represents a person, with a name and age. After creating at least 3 instances of the class, create a method that converts these instances into a dictionary so they can be saved to a JSON file named “exercicio7.json”. This method must be a static method of the class. Read the file after saving it.

````python
import json

class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  @staticmethod
  def to_dict(*args):
    dictionary = dict()
    for person in args:
      dictionary[person.name] = person.age
    return dictionary

person1 = Person('Carlos', 45)
person2 = Person('Maria', 67)
person3 = Person('Pedro', 23)

person_dict = Person.to_dict(person1, person2, person3)
text = json.dumps(person_dict, indent=4)
print(text)

with open('exercicio7.json', 'wt') as file:
  file.write(text)
````

8 - Based on the previous exercise, now create a class method that reads the generated file and returns the class instances as a list.

````python
import json

class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  @staticmethod
  def to_person():
    people = list()
    with open('exercicio7.json', 'rt') as file:
      read_file = file.read()
      dictionary = json.loads(read_file)
      for key in dictionary:
        person = Person(key, dictionary[key])
        people.append(person)
    return people

people = Person.to_person()

for person in people:
  print(person.name, person.age)
````

9 - Create an XML file. This XML file should have a root tag called Root. Inside this root, there can be multiple State tags with a name attribute. Each state can contain City tags, where the tag's text value should be the name of the city. Create a program that generates this file with some states and cities.



```
<Root>
    <State name="Rio Grande do Sul">
        <City>Santa Maria</City>
        <City>Porto Alegre</City>
        <City>Novo Hamburgo</City>
    </State>
    <State name="São Paulo">
        <City>Campinas</City>
        <City>Sorocaba</City>
    </State>
</Root>
```



````python
import xml.etree.ElementTree as xml
import os


def create_state(name, cities):
  state_element = xml.Element("State", attrib={'name': name})
  for city in cities:
    city_element = xml.SubElement(state_element, 'City')
    city_element.text = city
  return state_element

root = xml.Element("Root")

state_node = create_state("Rio Grande do Sul", ["Santa Maria", "Porto Alegre", "Novo Hamburgo"])
state_node2 = create_state("São Paulo", ["Sorocaba", "Campinas"])

root.append(state_node)
root.append(state_node2)

tree = xml.ElementTree(root)

with open('exercicio9.xml', 'wb') as files:
  tree.write(files)
````
