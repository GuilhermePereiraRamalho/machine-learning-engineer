```python
import pandas as pd
import urllib3
from bs4 import BeautifulSoup
```

# Searching for Date and Time on a Website

```python
def make_request(site):
  urllib3.disable_warnings(urllib3.exceptions.InsecureRequestWarning)
  manager = urllib3.PoolManager()
  return manager.request('GET', site, headers={'User-Agent': 'Mozilla/5.0'})

def read_site(site):
  response = make_request(site)
  return BeautifulSoup(response.data, 'html.parser')
```

# Searching for a Table on Wikipedia and Converting It into a DataFrame

```python
def read_countries():
  bs = read_site('https://pt.wikipedia.org/wiki/Lista_de_pa%C3%ADses_por_riqueza_total')
  table = bs.find('table', style='text-align: right')
  if (table == None):
    raise Exception("Table not found")

  body = table.find('tbody')
  if (body == None):
    raise("Table body not found")

  items = body.find_all('tr')

  countries = []
  for item in items:
    data = item.find_all('td')
    item_country = []
    for value in data:
      item_country.append(value.text.replace('\xa0', ''))
    if len(item_country) > 0:
      countries.append(item_country)
  return countries
```

```python
read_countries()
```

```python
data = pd.DataFrame(read_countries(), columns=['Position', 'Country', 'Total Wealth in Billions (USD)'])
data.set_index("Position")
```
